Exercício 2 — Ordenação com Cartas

A Resposta Teórica (Para colar em uma célula Markdown do seu Notebook):

1. Estado Inicial, Operações e Estado Final:
Estado Inicial: 13 cartas de espadas (1 a 13) alinhadas na mesa de forma aleatória.Operações Permitidas: Visualizar duas cartas por vez; trocar duas cartas de lugar usando apenas um "espaço temporário" (sua mão).Estado Final: As 13 cartas em ordem crescente (Ás até Rei).

2. O Algoritmo Passo a Passo (A Lógica):
Olho para a primeira carta da mesa e assumo que ela é a menor.Percorro o restante das cartas à direita olhando uma a uma. Sempre que achar uma carta menor que a que gravei na mente, atualizo minha memória de "onde está a menor".Ao chegar no final da mesa, pego a menor carta que encontrei e troco de lugar com a primeira carta da mesa. (Pronto, a 1ª posição está ordenada).Repito o processo começando da 2ª carta, depois da 3ª, até chegar na penúltima.

3. Qual é esse algoritmo clássico?
É o Selection Sort (Ordenação por Seleção), pois a cada rodada ele "seleciona" o menor elemento e o coloca na posição correta fazendo apenas uma troca física no final.

4. Custo em Notação Big O:
Para achar a 1ª carta, comparei 12 vezes. Para achar a 2ª, comparei 11 vezes... A soma disso dá $78$ comparações. Como temos dois laços (um andando pela mesa e outro procurando o menor), o número de comparações cresce de forma quadrática: $O(N^2)$. Já o número de trocas físicas é no máximo $N-1$, ou seja, $O(N)$.

Exercício 3 — Busca Linear

1. Dados do Problema:
Array: [2, 4, 6, 8, 10, 12, 13] (Tamanho $N = 7$)
Alvo da busca: 8

2. Simulação Passo a Passo:
Passo 1 (Índice 0): O valor 2 é igual a 8? -> Falso (1ª comparação).
Passo 2 (Índice 1): O valor 4 é igual a 8? -> Falso (2ª comparação).
Passo 3 (Índice 2): O valor 6 é igual a 8? -> Falso (3ª comparação).
Passo 4 (Índice 3): O valor 8 é igual a 8? -> Verdadeiro! (4ª comparação). 
Achamos o alvo!

3. Contagem Final e Justificativa:
Foram realizadas 4 comparações. A Busca Linear não "sabe" que o array está ordenado, então ela é obrigada a checar casa por casa, do índice 0 em diante, até topar com o número.

4. Relação com o tamanho do array ($N$):
O número de comparações cresce de forma diretamente proporcional (linear) ao tamanho da entrada e à posição do elemento. No pior cenário (buscar o 13 ou um número que não existe), o algoritmo faria exatamente 7 comparações (ou seja, $N$). Logo, a complexidade no pior caso é $O(N)$.

Exercício 4 — Busca Binária

1. Dados do Problema: 
Array: [2, 4, 6, 8, 10, 12, 13] (Ordenado, Tamanho $N = 7$)Índices: 0 a 6 Alvo: 8

2. Simulação Passo a Passo:
Estado Inicial: Ponteiro da esquerda (inicio = 0), ponteiro da direita (fim = 6).
Iteração 1: Calculamos o meio: (0 + 6) // 2 = 3 (Índice 3).
O valor no índice 3 é 8.
1ª Comparação: 8 == 8? -> Verdadeiro! O alvo foi encontrado.

3. Comparação com a Busca Linear: 
A Busca Binária encontrou o elemento com apenas 1 comparação (Melhor Caso), enquanto a Busca Linear precisou de 4 comparações.

4. Relação com a Notação Big O: 
A Busca Linear é $O(N)$ (descarta apenas 1 elemento por vez). A Busca Binária é $O(\log_2 N)$ porque descarta metade dos elementos restantes a cada erro. Se o array tivesse 1 milhão de elementos, a Busca Binária precisaria de no máximo umas 20 perguntas (pois $2^{20} \approx 1.000.000$), enquanto a Linear poderia precisar de 1 milhão de perguntas.

Exercício 5 — Otimização de Algoritmo 

1. Motivo da complexidade $O(N^2)$ no código original: A função possuía um laço for i e, dentro dele, um laço aninhado for j. Para cada um dos $N$ elementos do array, ela percorria todos os $N$ elementos de novo para contar. $N \times N = N^2$ operações. Em uma lista de 10.000 itens, isso geraria 100 milhões de iterações.

2. Estratégia de Otimização:Utilizamos uma estrutura de Tabela Hash (um dict em Python). Ao invés de laços aninhados, fazemos dois laços separados:O 1º laço percorre a lista uma vez ($O(N)$) e guarda quantas vezes cada número aparece no dicionário (a inserção no dict custa $O(1)$).O 2º laço percorre o dicionário ($O(N)$) apenas para achar o maior número que apareceu exatamente 1 vez.

3. Justificativa da Nova Complexidade:Somando dois laços separados: $O(N) + O(N) = O(2N)$. Pela regra do Big O, ignoramos a constante multiplicativa, resultando em uma complexidade linear final de $O(N)$.

```python

def maior_unico_linear(arr):
    frequencias = {} 

   
    for num in arr:
        
        frequencias[num] = frequencias.get(num, 0) + 1

    
    maior = None
    for num, contagem in frequencias.items():
        if contagem == 1:
            if (maior is None) or (num > maior):
                maior = num

    return maior


testes = [
    [4, 9, 2, 9, 4, 7, 2, 1], 
    [10, 10, 20, 20, 30, 30],
    [5, 3, 8, 12, 12, 8, 9]   
]

for caso in testes:
    print(f"Entrada: {caso} -> Maior Único: {maior_unico_linear(caso)}")
```

Exercício 6 — Tabuleiro de Xadrez e Grãos de Arroz

A cada iteração do laço, a variável graos_no_quadrado é multiplicada por 2. O número de iterações necessárias para alcançar a quantidade alvo $G$ cresce de forma logarítmica em relação a $G$.Por exemplo, para um alvo de 1.000 grãos, precisamos de cerca de 11 passos ($2^{10} = 1024$). Para 1.000.000 de grãos, precisamos de cerca de 21 passos ($2^{20} \approx 1.000.000$).Logo, a complexidade temporal em função da entrada qtd_graos é $O(\log G)$. 

```python
def encontrar_quadrado_xadrez(qtd_graos):
    if qtd_graos <= 0:
        return "A quantidade deve ser um número inteiro positivo.", None, 0

    graos_no_quadrado = 1 
    iteracoes = 0

   
    for quadrado in range(1, 65):
        iteracoes += 1
        
        
        if graos_no_quadrado >= qtd_graos:
            return quadrado, graos_no_quadrado, iteracoes
            
       
        graos_no_quadrado *= 2

    return "Valor excede o limite do tabuleiro (64 casas)", None, iteracoes

# Testes exigidos 
alvos_teste = [
    1,          
    8,          
    1000,      
    1000000,    
    10**20      
]

print(f"{'Alvo Desejado':<22} | {'Quadrado':<10} | {'Grãos na Casa':<22} | {'Iterações (Custo)'}")
print("-" * 75)

for alvo in alvos_teste:
    quad, graos, passos = encontrar_quadrado_xadrez(alvo)
    if graos is not None:
        print(f"{alvo:<22} | {quad:<10} | {graos:<22} | {passos}")
    else:
        print(f"{alvo:<22} | {quad} (Passos: {passos})")
```

Exercício 8 — Bubble Sort Bidirecional

```python
def cocktail_sort(arr):
    lista = arr[:] 
    esquerdo = 0
    direito = len(lista) - 1
    houve_troca = True

    while esquerdo < direito and houve_troca:
        houve_troca = False

       
        for i in range(esquerdo, direito):
            if lista[i] > lista[i + 1]:
                lista[i], lista[i + 1] = lista[i + 1], lista[i] 
                houve_troca = True
        
       
        direito -= 1 

        if not houve_troca:
            break 

        houve_troca = False

      
        for i in range(direito, esquerdo, -1):
            if lista[i - 1] > lista[i]:
                lista[i - 1], lista[i] = lista[i], lista[i - 1] 
                houve_troca = True
                
       
        esquerdo += 1 

    return lista


array_teste = [2, 3, 4, 5, 6, 7, 1]
print(f"Lista Original: {array_teste}")
print(f"Lista Ordenada (Cocktail Sort): {cocktail_sort(array_teste)}")
```

Exercícios 9 e 10 — O Bubble Sort Clássico

```python
# ================= Ex. 9 (Bubble Sort Numérico) =================
def bubble_sort_numerico(arr):
    lista = arr[:]
    n = len(lista)

    
    for i in range(n - 1):
        
        for j in range(0, n - 1 - i):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista


print("Ex 9 - Desordenada:", [64, 34, 25, 12, 22] ,"->", bubble_sort_numerico([64, 34, 25, 12, 22]))


# ================= Ex. 10 (Bubble Sort para Strings) =================
def bubble_sort_strings(palavras):
    lista = palavras[:]
    n = len(lista)

    for i in range(n - 1):
        for j in range(0, n - 1 - i):
            
            if lista[j].lower() > lista[j + 1].lower():
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista


frutas = ["banana", "abacaxi", "Laranja", "uva", "Manga"]
print("Ex 10 - Frutas Ord:", frutas, "->", bubble_sort_strings(frutas))
```

Exercícios 11 e 12 — A "Cirurgia" do Bubble Sort

```python
# Exercício 11 - Tradicional e "Burro" (Sem critério de parada)
def bubble_sort_instrumentado(arr):
    lista = arr[:]
    n = len(lista)
    comparacoes = 0
    trocas = 0

    for i in range(n - 1):
        for j in range(0, n - 1 - i):
            comparacoes += 1
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                trocas += 1

    return comparacoes

# Exercício 12 - Otimizado e "Inteligente" (Com flag houve_troca)
def bubble_sort_otimizado(arr):
    lista = arr[:]
    n = len(lista)
    comparacoes = 0

    for i in range(n - 1):
        houve_troca = False 
        for j in range(0, n - 1 - i):
            comparacoes += 1
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                houve_troca = True
                
        if not houve_troca:
            break 

    return comparacoes


print("CUSTO DO BUBBLE SORT NUMA LISTA JÁ ORDENADA DE 100 ITENS:")
lista_perfeita = list(range(100)) 

custo_ruim = bubble_sort_instrumentado(lista_perfeita)
custo_bom = bubble_sort_otimizado(lista_perfeita)

print(f"-> Bubble Sort Clássico (Ex 11) fez: {custo_ruim} comparações inúteis.")
print(f"-> Bubble Sort Otimizado (Ex 12) fez: {custo_bom} comparações.")
print(f"ECONOMIA DE CPU: {custo_ruim - custo_bom} operações ({(1 - custo_bom/custo_ruim)*100:.1f}%)!")
```

No Exercício 11 (Clássico), mesmo a lista já estando em ordem, a complexidade de Melhor Caso é de espantosos $\Omega(N^2)$.

No Exercício 12 (Otimizado), ao adicionar a flag houve_troca, o algoritmo percebe logo na primeira passada que a lista está organizada e para. Com isso, melhoramos drasticamente o limite inferior do algoritmo (Melhor Caso), transformando-o de $\Omega(N^2)$ para a incrível marca de $\Omega(N)$. O limite superior (Pior Caso) se mantém em $O(N^2)$.

Exercício 13 — O Grande Combate: Busca Linear vs. Busca Binária

```python
import random



def linear_search_count(arr, target):
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

def binary_search_count(sorted_arr, target):
    comparisons = 0
    left = 0
    right = len(sorted_arr) - 1

    while left <= right:
        mid = (left + right) // 2
        comparisons += 1
        
        if sorted_arr[mid] == target:
            return mid, comparisons
        elif target < sorted_arr[mid]:
            right = mid - 1
        else:
            left = mid + 1

    return -1, comparisons

def bubble_sort_count(arr):
    copia = arr[:]
    comparisons = 0
    for i in range(len(copia) - 1):
        for j in range(0, len(copia) - 1 - i):
            comparisons += 1
            if copia[j] > copia[j + 1]:
                copia[j], copia[j + 1] = copia[j + 1], copia[j]
    return copia, comparisons


N = 31
arr_sorted = list(range(10, 10 + N * 2, 2))  
arr_reverse = arr_sorted[::-1]
arr_random = arr_sorted[:]
random.seed(42)
random.shuffle(arr_random)

listas_teste = [
    ("Ordenada", arr_sorted, True),
    ("Decrescente", arr_reverse, False),
    ("Embaralhada", arr_random, False)
]

print(f"{'Cenário':<14} | {'Alvo':<12} | {'Linear':<8} | {'Binária':<8} | {'Sort':<6} | {'Total (Bin+Sort)'}")
print("-" * 75)

for nome_lista, lista, ja_ordenada in listas_teste:
    alvos = [("Primeiro", lista[0]), ("Meio", lista[N // 2]), ("Ausente", -1)]

    for nome_alvo, target in alvos:
        
        _, comp_linear = linear_search_count(lista, target)

        
        if ja_ordenada:
            _, comp_bin = binary_search_count(lista, target)
            comp_sort = 0
            total_bin = comp_bin
        else:
            lista_ord, comp_sort = bubble_sort_count(lista)
            _, comp_bin = binary_search_count(lista_ord, target)
            total_bin = comp_sort + comp_bin

        print(f"{nome_lista:<14} | {nome_alvo:<12} | {comp_linear:<8} | {comp_bin:<8} | {comp_sort:<6} | {total_bin}")
```

1. Quais são os pré-requisitos de cada busca?A Busca Linear não exige nenhum pré-requisito; ela aceita dados totalmente desordenados. Já a Busca Binária exige estritamente que a lista permita acesso por índice e esteja previamente ordenada.

2. Qual é o Big O da busca linear e da busca binária?A Busca Linear possui complexidade de pior caso $O(N)$. A Busca Binária tem complexidade logarítmica $O(\log_2 N)$, sendo imensamente superior para grandes volumes.

3. Quando faz sentido pagar o custo do Bubble Sort para usar a Busca Binária?Se formos realizar apenas uma única consulta, não faz sentido. O custo para ordenar no Bubble Sort é $O(N^2)$. Pagar $O(N^2)$ para economizar na busca é um mau negócio comparado a fazer uma única Busca Linear que custa só $O(N)$.Porém, se fomos construir um sistema que fará múltiplas consultas (milhares) sobre a mesma base de dados estática, vale muito a pena pagar o custo da ordenação uma única vez, pois as milhares de buscas seguintes serão resolvidas quase instantaneamente em $O(\log N)$, amortizando o custo inicial.

4. Relação empírica observada:Na tabela gerada ($N=31$), buscar um elemento ausente ordenado levou 31 passos na Linear e apenas 5 na Binária. Contudo, nas listas desordenadas, fomos obrigados a gastar $465$ operações de ordenação prévia. O total de "Sort + Busca Binária" subiu para incríveis $470$ passos para achar 1 único elemento, contra apenas 31 da Busca Linear.

Exercício 14 — Refatoração e Clean Code

```python

def f(a, t):
    c = 0; i = 0; j = len(a) - 1; r = -1
    while i <= j:
        m = (i + j) // 2
        c = c + 1
        if a[m] == t:
            r = m
            break
        c = c + 1
        if t < a[m]:
            j = m - 1
        else:
            i = m + 1
    return r, c

def h(arr):
    b = arr[:]
    n = len(b)
    k = 0
    for last in range(n - 1, 0, -1):
        for p in range(last):
            k += 1
            if b[p] > b[p + 1]:
                tmp = b[p]; b[p] = b[p + 1]; b[p + 1] = tmp
    return b, k



def busca_binaria_contagem(lista_ordenada, alvo):
    """ Busca binária com contagem - Complexidade: O(log N) """
    total_comparacoes = 0
    indice_esquerdo = 0
    indice_direito = len(lista_ordenada) - 1
    indice_encontrado = -1

    while indice_esquerdo <= indice_direito:
        indice_meio = (indice_esquerdo + indice_direito) // 2
        
       
        total_comparacoes += 1
        if lista_ordenada[indice_meio] == alvo:
            indice_encontrado = indice_meio
            break
            
      
        total_comparacoes += 1
        if alvo < lista_ordenada[indice_meio]:
            indice_direito = indice_meio - 1
        else:
            indice_esquerdo = indice_meio + 1

    return indice_encontrado, total_comparacoes


def bubble_sort_contagem(lista_original):
    """ Bubble Sort com contagem - Complexidade: O(N²) """
    copia_lista = lista_original[:]
    tamanho_lista = len(copia_lista)
    total_comparacoes = 0

    for limite_direito in range(tamanho_lista - 1, 0, -1):
        for indice_atual in range(limite_direito):
            total_comparacoes += 1
            
            
            if copia_lista[indice_atual] > copia_lista[indice_atual + 1]:
                temporario = copia_lista[indice_atual]
                copia_lista[indice_atual] = copia_lista[indice_atual + 1]
                copia_lista[indice_atual + 1] = temporario

    return copia_lista, total_comparacoes



print("--- TESTE DE EQUIVALÊNCIA DE REFATORAÇÃO ---")
dados = [42, 15, 8, 99, 23, 4, 16, 15]


res_orig_sort = h(dados)
res_refat_sort = bubble_sort_contagem(dados)
assert res_orig_sort == res_refat_sort, "Erro: As ordenações estão diferentes!"
print(f"[OK] Bubble Sort refatorado manteve a fidelidade. Total comparações: {res_refat_sort[1]}")


lista_ordenada = res_refat_sort[0]
alvo_teste = 23

res_orig_busca = f(lista_ordenada, alvo_teste)
res_refat_busca = busca_binaria_contagem(lista_ordenada, alvo_teste)
assert res_orig_busca == res_refat_busca, "Erro: As buscas estão diferentes!"
print(f"[OK] Busca Binária refatorada manteve a fidelidade para o alvo {alvo_teste}!")
```
